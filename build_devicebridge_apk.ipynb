# 📱 DeviceBridge APK Builder (Google Colab)

Notebook ini digunakan untuk mengompilasi (build) aplikasi **DeviceBridge Android APK** secara otomatis di Google Colab tanpa memerlukan komputer/laptop canggih atau penyimpanan di HP Anda.

---  
### 🚀 Cara Menggunakan:
1. Ganti `REPO_URL` dengan URL repositori GitHub DeviceBridge Anda.
2. Jalankan cell di bawah ini dengan menekan tombol **Play (▶)**.
3. Tunggu 1-2 menit hingga proses build selesai, file APK akan **otomatis terunduh** ke HP/Laptop Anda!

```python
# @title 🔨 Build & Download DeviceBridge APK
import os
from google.colab import files

# 1. Masukkan URL Repo GitHub DeviceBridge Anda di sini
REPO_URL = "https://github.com/YOUR_USERNAME/devicebridge.git"  # @param {type:"string"}

print("📥 Cloning repository...")
!git clone {REPO_URL} repo_project || echo "Repo already cloned or Zip uploaded."

if os.path.exists("repo_project"):
    %cd repo_project

# 2. Sinkronkan assets web-client ke android assets
print("📂 Syncing web-client assets...")
!mkdir -p android-app/app/src/main/assets/web
!cp -r web-client/* android-app/app/src/main/assets/web/

# 3. Beri izin eksekusi gradlew
%cd android-app
!chmod +x gradlew

# 4. Jalankan kompilasi APK Release
print("⚙️ Compiling APK with Gradle...")
!./gradlew assembleRelease --no-daemon

# 5. Unduh file APK jika berhasil
apk_path = 'app/build/outputs/apk/release/app-release-unsigned.apk'
if os.path.exists(apk_path):
    print("🎉 BUILD BERHASIL! Mengunduh file APK...")
    files.download(apk_path)
else:
    print("❌ Build gagal. Silakan periksa log error di atas.")
```
